```python
import cmath
import math


z = complex(4,4)
n = 3

if n <= 0:
  raise ValueError(" n must be a positive integer. ")
else:
  r, theta = cmath.polar(z)

  root_r = r**(1/n)
  roots = []

  print(f"--- The {n} roots of {z} ---")

  for k in range(n):
    root_theta = (theta + 2*math.pi*k)/n
    root_cartesian = cmath.rect(root_r, root_theta)

    roots.append(root_cartesian)

    print(f"\nRoot {k+1}:")
    print(
        f"  Cartesian form: {root_cartesian.real:.4f} + {root_cartesian.imag:.4f}j"
      )
    print(
        f"  Polarform: Magnitude = {root_r:.4f}, Angle = {root_theta:.4f} rad"
      )
```

```python
import numpy as np
import matplotlib.pyplot as plt
import cmath

angle = np.linspace(0,2*np.pi,50)

x_coord = []
y_coord = []

for x in angle:

  matrix1 = np.array([[np.cos(x) , -np.sin(x)],
                      [np.sin(x) , np.cos(x)]])

  Y = np.array([[2], [5]])
  product = matrix1 @ Y

  x_coord.append( product[0] )
  y_coord.append( product[1] )

plt.scatter(x_coord, y_coord)
plt.title("Rotated Vector Path")

plt.show()
```

```python
#Translation

def translate(z,dx,dy):
    real_part = z.real + dx
    imag_part = z.imag +dy
    return real_part + 1j*imag_part

z = complex(3,2)
translate(z,2,-1)
```

```python
#Scaling

def scale(z,factor):
    real_part = z.real*factor
    imag_part = z.imag*factor
    return real_part + 1j*imag_part

z = complex(3,2)
scale(z,2.5)
```

```python
#Reflection

def reflection(z):
    real_part = z.real
    imag_part = -z.imag
    return real_part + 1j*imag_part

reflection( 5+1j )
```

```python
# Fourier transform

import numpy as np
import matplotlib.pyplot as plt

#input
N =100
t = np.linspace(0,1,N)
f = 5

signal = np.sin(2*np.pi * f * t)

result = np.fft.fft(signal)
freq = np.fft.fftfreq(N, t[1] - t[0])

plt.plot(freq[:N//2], np.abs(result[:N//2]), color = "c")
plt.xlabel("Frequency(Hz)")
plt.ylabel("Magnitude")
plt.grid()
plt.show()
```

```python
# Inverse Fourier Transform
import numpy as np
import matplotlib.pyplot as plt

N = 100
t = np.linspace(0,1,N)
f = 5

signal = np.sin(2*np.pi * f * t)

result = np.fft.fft(signal)

#inverse ft
inverse = np.fft.ifft(result)

plt.plot(t, np.real(inverse), label = "Inverse Fourier Transform")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.title("Inverse Fourier Transform")
plt.grid()
plt.show()
```
